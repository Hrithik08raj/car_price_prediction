```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
import matplotlib as mpl
mpl.style.use('ggplot')
```

```python
car = pd.read_csv('quikr_car.csv')
```

```python
car.head()
```

```python
car.info()
```

```python
# Creating backup copy:
backup = car.copy()
```

```python
# Data cleaning:
# 1 - Year has many non-year values:
car = car[car['year'].str.isnumeric()]
car.head()
```

```python
# 2 - Year is in object,change in integer:
car['year'] = car['year'].astype(int)
```

```python
car.info()
```

```python
# 3 - Remove unrelevant values from price column:
car = car[car['Price'] != 'Ask For Price']
car.head()
```

```python
#  4 - Price has commas in its prices and is in object:
car['Price'] = car['Price'].str.replace(',','').astype(int)
```

```python
car.head()
```

```python
car.info()
```

```python
# 5 - kms_driven has object values with kms at last.
car['kms_driven'] = car['kms_driven'].str.split().str.get(0).str.replace(',','')
```

```python
# 6 - It has nan values and two rows have 'Petrol' in them:
car = car[car['kms_driven'].str.isnumeric()]
car.head()
```

```python
car['kms_driven'] = car['kms_driven'].astype(int)
```

```python
car.info()
```

```python
# 7 - fuel_type has nan values:
car =  car[~car['fuel_type'].isna()]
car.head()
```

```python
car.shape
```

```python
# 8 - Changing car names. Keeping only the first three words:
car['name'] = car['name'].str.split().str.slice(start = 0 , stop = 3).str.join(' ')
```

```python
car.head()
```

```python
car = car.reset_index(drop = True)
```

```python
car.head()
```

```python
# Convert the cleaned data into csv file:
car.to_csv('Cleaned_Car_data.csv')
```

```python
car.info()
```

```python
# Extracting data:
X = car[['name','company','year','kms_driven','fuel_type']]
y = car['Price']
```

```python
X.head()
```

```python
y.head()
```

```python
# Applying train test split:
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2)
```

```python
# Import models:
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
from sklearn.metrics import r2_score
```

```python
# Creating an OneHotEncoder object to contain all the possible categories:
ohe = OneHotEncoder()
ohe.fit(X[['name','company','fuel_type']])
```

```python
# Creating a column transformer to transform categorical columns:
column_trans = make_column_transformer((OneHotEncoder(categories = ohe.categories_), ['name','company','fuel_type']),remainder = 'passthrough')
```

```python
# Linear Regression Model:
lr = LinearRegression()
```

```python
# Making a pipeline:
pipe = make_pipeline(column_trans,lr)
```

```python
# Fitting the model:
pipe.fit(X_train,y_train)
```

```python
y_pred = pipe.predict(X_test)
```

```python
# Checking r2 score:
r2_score(y_test,y_pred)
```

```python
# Finding the model with a random state of TrainTestSplit where the model was found to give almost 0.92 as r2_score:
scores = []
for i in range(1000):
  X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.1, random_state = i)
  lr = LinearRegression()
  pipe = make_pipeline(column_trans,lr)
  pipe.fit(X_train,y_train)
  y_pred = pipe.predict(X_test)
  scores.append(r2_score(y_test,y_pred))
```

```python
np.argmax(scores)
```

```python
scores[np.argmax(scores)]
```

```python
pipe.predict(pd.DataFrame(columns = X_test.columns, data = np.array(['Maruti Suzuki Swift','Maruti',2019,100,'Petrol']).reshape(1,5)))
```

```python
# The best model is found at a certain random state:
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.1,random_state = np.argmax(scores))
lr = LinearRegression()
pipe = make_pipeline(column_trans,lr)
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
r2_score(y_test,y_pred)
```
